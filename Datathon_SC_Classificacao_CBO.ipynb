```python

import pandas as pd

# Abrir o arquivo do dicionário
caminho = '/content/Dicionario_de_dados (1).xlsx'

# Identificar todas as abas existentes
excel = pd.ExcelFile(caminho)

print("Abas encontradas no dicionário:\n")

for aba in excel.sheet_names:
    print("-", aba)
```

```python

# Abrir a aba que contém os códigos CBO
tabela_cbo = pd.read_excel(
    caminho,
    sheet_name='cbo2002ocupação'
)

# Mostrar os nomes das colunas
print("Colunas da tabela:")
print(tabela_cbo.columns.tolist())

# Mostrar as primeiras 10 linhas
print("\nPrimeiras 10 ocupações:")
display(tabela_cbo.head(10))
```

```python

caminho_caged = '/content/Novo_Caged_MOV_Limpo.csv'

# Ler apenas as primeiras linhas do arquivo original
with open(caminho_caged, 'r', encoding='utf-8-sig') as arquivo:
    for i in range(3):
        print(arquivo.readline())
```

```python

# Carregar a base completa do Novo CAGED
caged = pd.read_csv(
    '/content/Novo_Caged_MOV_Limpo.csv',
    sep=';',
    encoding='utf-8-sig',
    dtype={'cbo2002ocupação': 'string'},
    low_memory=False
)

# Conferir o tamanho da base
print("Quantidade de linhas:", len(caged))
print("Quantidade de colunas:", len(caged.columns))

# Conferir os primeiros registros
display(caged.head())
```

```python

# Contar os códigos CBO diferentes
total_cbo = caged['cbo2002ocupação'].nunique()

# Contar registros sem código CBO
sem_cbo = caged['cbo2002ocupação'].isna().sum()

print("Total de códigos CBO únicos:", total_cbo)
print("Registros sem código CBO:", sem_cbo)

# Mostrar alguns códigos encontrados
print("\nExemplos de códigos CBO:")
display(
    caged['cbo2002ocupação']
    .dropna()
    .drop_duplicates()
    .head(15)
)
```

```python

# Contar quantos códigos CBO diferentes existem
total_cbo = caged['cbo2002ocupação'].nunique()

# Verificar registros sem código
sem_cbo = caged['cbo2002ocupação'].isna().sum()

print("Total de linhas:", len(caged))
print("Total de CBOs únicos:", total_cbo)
print("Registros sem CBO:", sem_cbo)

# Mostrar alguns códigos diferentes
codigos = caged['cbo2002ocupação'].dropna().drop_duplicates()

print("\nPrimeiros 20 códigos CBO:")
display(codigos.head(20))
```

```python

# 1. Preparar os códigos do CAGED
cbo_caged = pd.DataFrame({
    'Código': caged['cbo2002ocupação'].dropna().unique()
})

# 2. Padronizar os códigos com 6 dígitos
cbo_caged['Código'] = (
    cbo_caged['Código'].astype(str).str.strip().str.zfill(6)
)

# 3. Preparar o dicionário
dicionario_cbo = tabela_cbo.copy()

dicionario_cbo['Código'] = (
    dicionario_cbo['Código']
    .astype(str)
    .str.replace(r'\.0$', '', regex=True)
    .str.strip()
    .str.zfill(6)
)

# 4. Relacionar os códigos com as profissões
cbo_com_descricoes = cbo_caged.merge(
    dicionario_cbo[['Código', 'Descrição']],
    on='Código',
    how='left',
    indicator=True
)

# 5. Conferir os resultados
encontrados = (cbo_com_descricoes['_merge'] == 'both').sum()
nao_encontrados = (cbo_com_descricoes['_merge'] == 'left_only').sum()

print("CBOs encontrados no dicionário:", encontrados)
print("CBOs não encontrados:", nao_encontrados)

print("\nExemplos de códigos relacionados:")
display(cbo_com_descricoes.head(15))

print("\nCódigos que não foram encontrados:")
display(
    cbo_com_descricoes.loc[
        cbo_com_descricoes['_merge'] == 'left_only',
        ['Código']
    ].head(20)
)
```

```python

cbo_nao_encontrados = cbo_com_descricoes[
    cbo_com_descricoes['_merge'] == 'left_only'
][['Código']].copy()

print("Códigos que precisam ser investigados:")
display(cbo_nao_encontrados)

cbo_nao_encontrados.to_csv(
    '/content/CBOs_nao_encontrados.csv',
    index=False,
    sep=';',
    encoding='utf-8-sig'
)

print("Lista salva com sucesso!")
```

```python

# Verificar se existem códigos repetidos no dicionário
duplicados = dicionario_cbo[
    dicionario_cbo['Código'].duplicated(keep=False)
].sort_values('Código')

print("Quantidade de linhas com códigos duplicados:",
      len(duplicados))

display(duplicados)

```

```python
cbo_nao_encontrados = cbo_com_descricoes[
    cbo_com_descricoes['_merge'] == 'left_only'
][['Código']].copy()

display(cbo_nao_encontrados)
```

```python
cbo_classificacao = cbo_com_descricoes[
    cbo_com_descricoes['_merge'] == 'both'
][['Código', 'Descrição']].copy()

print("Total de CBOs para classificação:", len(cbo_classificacao))

display(cbo_classificacao.head(10))
```

```python
print("Códigos duplicados:", cbo_classificacao['Código'].duplicated().sum())

print("Descrições vazias:", cbo_classificacao['Descrição'].isna().sum())

print("Total de ocupações:", len(cbo_classificacao))
```

```python
cbo_classificacao.to_csv(
    '/content/CBOs_para_classificacao.csv',
    index=False,
    sep=';',
    encoding='utf-8-sig'
)

print("Arquivo salvo com sucesso!")
```

```python
# Contar quantas vezes cada CBO aparece no CAGED
frequencia_cbo = (
    caged['cbo2002ocupação']
    .value_counts()
    .rename_axis('Código')
    .reset_index(name='Quantidade de registros')
)

# Adicionar as frequências à tabela de ocupações
cbo_com_frequencia = cbo_classificacao.merge(
    frequencia_cbo,
    on='Código',
    how='left'
)

# Ordenar da maior para a menor frequência
cbo_com_frequencia = cbo_com_frequencia.sort_values(
    by='Quantidade de registros',
    ascending=False
)

print("Total de ocupações:", len(cbo_com_frequencia))

display(cbo_com_frequencia.head(30))
```
